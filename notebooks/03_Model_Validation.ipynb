```python
eeg_bands = {'Delta': (1, 4),
             'Theta': (4, 8),
             'Alpha': (8, 12),
             'Beta': (12, 30),
             'Gamma': (30, 45)}
```

```python
mean = np.load("/content/drive/MyDrive/Project/mean.npy")
std = np.load("/content/drive/MyDrive/Project/std.npy")
```

```python
def data_transform(data):
  result = []
  eeg_data = np.array(data)
  t_data = np.array(eeg_data[-1250:]).T
  fs = 120

  for k in range(1,9):
    eeg_band_fft = dict()
    sp_bands = np.absolute(np.fft.fft(t_data[k]))
    freq_bands = np.fft.fftfreq(t_data[k].shape[-1], 1.0/fs)

    for band in eeg_bands:
      freq_ix = np.where((freq_bands >= eeg_bands[band][0]) & (freq_bands <= eeg_bands[band][1]))[0]
      eeg_band_fft[band] = np.mean(sp_bands[freq_ix])

    eeg_band_arr = []
    eeg_band_column = ['Delta', 'Theta', 'Alpha', 'Beta', 'Gamma']
    for j in eeg_band_column:
      eeg_band_arr.append(eeg_band_fft[j])

    result.append(eeg_band_arr)

  result = np.array(result).reshape(1, -1)[0]
  # result = ((result - mean) / std).reshape(1, -1)[-1]

  return result
```
